```python
from trading.model_tools import ta_transform
import trading.model_tools as mt
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# %pip install ipympl
# %matplotlib ipympl

def shannon_entropy(series: pd.Series, bins: int = 50) -> float:
    values = series.dropna().to_numpy()
    counts, _ = np.histogram(values, bins=bins)
    probs = counts[counts > 0] / counts.sum()
    return -np.sum(probs * np.log2(probs))

data = mt.fetch_data(
    symbols=["SOL-USDT", "BTC-USDT"],
    days=730,
    interval="1h",
    age_days=0,
    data_source="binance",
    cache_expiry_hours=-1,
    verbose=True,
)

shift_by = "rsi"
features_original = ta_transform(data, add_ticker="BTC-USDT")
# for col in features_original.columns:
#     features_original[col] = np.fft.fft(features_original[col].to_numpy())
shifted_features = features_original.sub(features_original[shift_by], axis=0)

corr_matrix = np.round(features_original.corr(), 2)
shifted_corr_matrix = np.round(shifted_features.corr(), 2)

X_key = "add_ret_5"
y_key = "vol_ratio"
```

```python
entropy_original = {
    col: shannon_entropy(features_original[col], bins=50)
    for col in features_original.columns
}

entropy_shifted = {
    col: shannon_entropy(shifted_features[col], bins=50)
    for col in shifted_features.columns
}

features = list(entropy_original.keys())
orig_values = [entropy_original[f] for f in features]
shift_values = [entropy_shifted[f] for f in features]

x = np.arange(len(features))

fig, ax = plt.subplots(figsize=(14, 6))
ax.bar(x - 0.35 / 2, orig_values, 0.35, label="original")
ax.bar(x + 0.35 / 2, shift_values, 0.35, label="shifted")

ax.set_xticks(x)
ax.set_xticklabels(features, rotation=90)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix, annot=True)
plt.title("Original Features CM")
plt.show()

plt.figure(figsize=(12,8))
sns.heatmap(shifted_corr_matrix, annot=True)
plt.title("Shifted Features CM")
plt.show()

plt.figure(figsize=(12,8))
sns.heatmap(shifted_corr_matrix-corr_matrix, annot=True)
plt.title("Features CM Diff")
plt.show()
```

```python
plt.plot(features_original[X_key], features_original[y_key], marker = 'o', linestyle='None', label='original')
plt.plot(shifted_features[X_key], shifted_features[y_key], marker = 'o', linestyle='None', label='shifted')
plt.legend()
plt.show()
```
